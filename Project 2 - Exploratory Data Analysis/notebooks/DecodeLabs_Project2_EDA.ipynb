# DecodeLabs Internship — Project 2: Exploratory Data Analysis
This notebook analyzes `Final Dataset.xlsx` using pandas, NumPy and Matplotlib.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
df = pd.read_excel('../data/Final Dataset.xlsx', sheet_name='Cleaned_Data')
print('Shape:', df.shape)
display(df.head())
```

```python
# Basic statistics
display(df.describe(include='all').T)
display(df.select_dtypes(include=np.number).median().rename('median').to_frame())
```

```python
# Missing values and duplicates
display(df.isna().sum().sort_values(ascending=False).to_frame('missing'))
print('Duplicate rows:', df.duplicated().sum())
```

```python
# IQR outlier counts
for c in df.select_dtypes(include=np.number).columns:
    q1, q3 = df[c].quantile([.25,.75]); iqr=q3-q1
    lo, hi=q1-1.5*iqr, q3+1.5*iqr
    print(c, 'outliers:', int(((df[c]<lo)|(df[c]>hi)).sum()))
```

```python
if 'TotalPrice' in df.columns:
    df['TotalPrice'].plot(kind='hist', bins=30, title='Distribution of Total Price')
    plt.xlabel('Total Price'); plt.show()
if 'OrderStatus' in df.columns:
    df['OrderStatus'].value_counts().plot(kind='bar', title='Order Status Distribution')
    plt.xlabel('Order Status'); plt.ylabel('Orders'); plt.show()
```
